## 1. Project Objective

The objective of this project is to develop a **Hybrid ML-Based Intrusion Detection System (IDS)** capable of detecting network attacks using machine learning and deep learning techniques.

The proposed system will combine **FT-Transformer-based supervised classification** with **Autoencoder-based anomaly detection** to improve the detection of both known and potentially unseen network attacks.

The ML module will cover dataset preparation, preprocessing, exploratory data analysis, baseline model development, FT-Transformer implementation, Autoencoder implementation, hybrid model development, evaluation, and integration with the backend inference system.

## 2. Datasets

The project will use the following network intrusion detection datasets:

### CICIDS2017

CICIDS2017 will be used for network traffic analysis and supervised intrusion detection experiments. The dataset contains benign traffic along with multiple categories of network attacks.

### CIC-IoT2023

CIC-IoT2023 will be used to evaluate intrusion detection in an IoT-oriented network environment and provides additional attack scenarios for experimentation.

**Detailed dataset analysis, feature inspection, cleaning and preprocessing will be performed in the upcoming weeks.**

## 3. Planned ML Workflow

The planned ML workflow for the project is:

```text
CICIDS2017 / CIC-IoT2023
            ↓
      Data Loading
            ↓
     Data Cleaning
            ↓
    Encoding & Scaling
            ↓
           EDA
            ↓
 Train / Validation / Test Split
            ↓
      Baseline Models
            ↓
       FT-Transformer
            ↓
        Autoencoder
            ↓
    Model Comparison
            ↓
        Hybrid Model
            ↓
       Model Evaluation
            ↓
        ML Inference
            ↓
    Backend / API Integration
            ↓
      Prediction + Alert
```

This workflow will be implemented incrementally throughout the project.


## 5. Evaluation Metrics

Additional analysis such as confusion matrices and attack-wise performance will be considered during model evaluation.
