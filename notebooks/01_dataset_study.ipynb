```python
import pandas as pd
from pathlib import Path
```

```python
CICIDS_PATH = Path(
    r"D:\Project-root\SKIT-IOT-2023-2027-F-031\data\raw\CICIDS2017"
)

CIC_IOT_PATH = Path(
    r"D:\Project-root\SKIT-IOT-2023-2027-F-031\data\raw\CIC-IoT2023"
)
```

```python
cicids_files = list(CICIDS_PATH.glob("*.csv"))

print("Number of CSV files:", len(cicids_files))

for file in cicids_files:
    print(file.name)
```

```python
dataset_summary = []

for file in cicids_files:
    df = pd.read_csv(file)

    dataset_summary.append({
        "File": file.name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Label_Column": df.columns[-1]
    })

cicids_summary = pd.DataFrame(dataset_summary)

cicids_summary
```

```python
iot_files = list(CIC_IOT_PATH.glob("*/*.csv"))

print("Number of CSV files:", len(iot_files))

for file in iot_files:
    print(file.name)
```

```python
iot_summary = []

for file in iot_files:
    df = pd.read_csv(file)

    iot_summary.append({
        "File": file.name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Label_Column": df.columns[-1]
    })

iot_summary = pd.DataFrame(iot_summary)

iot_summary
```
